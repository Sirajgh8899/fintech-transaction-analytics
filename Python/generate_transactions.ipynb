```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Environment ready!")
```

```python
# Make our synthetic data reproducible
np.random.seed(42)

# Project configuration
NUM_CUSTOMERS = 5_000

countries = [
    "Germany",
    "France",
    "Netherlands",
    "Spain",
    "Italy",
    "Belgium"
]

# Generate customer data
customers = pd.DataFrame({
    "customer_id": [
        f"CUST_{i:05d}"
        for i in range(1, NUM_CUSTOMERS + 1)
    ],

    "country": np.random.choice(
        countries,
        size=NUM_CUSTOMERS,
        p=[0.25, 0.20, 0.15, 0.15, 0.15, 0.10]
    ),

    "customer_age": np.random.randint(
        18,
        70,
        size=NUM_CUSTOMERS
    ),

    "account_age_days": np.random.randint(
        30,
        1500,
        size=NUM_CUSTOMERS
    )
})

customers.head()
```

```python
print("Dataset shape:", customers.shape)

print("\nCustomers by country:")
print(customers["country"].value_counts())

print("\nMissing values:")
print(customers.isnull().sum())
```

```python
NUM_MERCHANTS = 50

merchant_categories = [
    "E-commerce",
    "Groceries",
    "Travel",
    "Entertainment",
    "Electronics",
    "Food & Dining",
    "Fashion",
    "Digital Services"
]

merchant_names = [
    f"Merchant_{i:03d}"
    for i in range(1, NUM_MERCHANTS + 1)
]

merchants = pd.DataFrame({
    "merchant_id": [
        f"MER_{i:03d}"
        for i in range(1, NUM_MERCHANTS + 1)
    ],

    "merchant_name": merchant_names,

    "merchant_category": np.random.choice(
        merchant_categories,
        size=NUM_MERCHANTS
    ),

    "merchant_country": np.random.choice(
        countries,
        size=NUM_MERCHANTS
    )
})

merchants.head(10)

```

```python
NUM_TRANSACTIONS = 100_000

# Transaction IDs
transaction_ids = [
    f"TXN_{i:06d}"
    for i in range(1, NUM_TRANSACTIONS + 1)
]

# Randomly assign customers and merchants
customer_ids = np.random.choice(
    customers["customer_id"],
    size=NUM_TRANSACTIONS
)

merchant_ids = np.random.choice(
    merchants["merchant_id"],
    size=NUM_TRANSACTIONS
)

# Generate timestamps across 2025
start_date = pd.Timestamp("2025-01-01")
end_date = pd.Timestamp("2025-12-31 23:59:59")

random_seconds = np.random.randint(
    0,
    int((end_date - start_date).total_seconds()),
    size=NUM_TRANSACTIONS
)

timestamps = start_date + pd.to_timedelta(random_seconds, unit="s")

# Generate transaction amounts
amounts = np.round(
    np.random.lognormal(
        mean=3.8,
        sigma=0.9,
        size=NUM_TRANSACTIONS
    ),
    2
)

# Payment methods
payment_methods = np.random.choice(
    [
        "Credit Card",
        "Debit Card",
        "Digital Wallet",
        "Bank Transfer"
    ],
    size=NUM_TRANSACTIONS,
    p=[0.40, 0.30, 0.20, 0.10]
)

# Device types
device_types = np.random.choice(
    [
        "Mobile",
        "Desktop",
        "Tablet"
    ],
    size=NUM_TRANSACTIONS,
    p=[0.55, 0.35, 0.10]
)

# Transaction types
transaction_types = np.random.choice(
    [
        "Purchase",
        "Subscription",
        "Refund"
    ],
    size=NUM_TRANSACTIONS,
    p=[0.82, 0.13, 0.05]
)

transactions = pd.DataFrame({
    "transaction_id": transaction_ids,
    "customer_id": customer_ids,
    "merchant_id": merchant_ids,
    "transaction_timestamp": timestamps,
    "amount": amounts,
    "transaction_type": transaction_types,
    "payment_method": payment_methods,
    "device_type": device_types
})

transactions.head()
```

```python
print("Transaction dataset shape:", transactions.shape)

print("\nFirst 5 transactions:")
display(transactions.head())

print("\nAmount statistics:")
display(transactions["amount"].describe())

print("\nPayment method distribution:")
print(transactions["payment_method"].value_counts(normalize=True).round(3))
```

```python
# Base failure probability by payment method
failure_probability = transactions["payment_method"].map({
    "Credit Card": 0.05,
    "Debit Card": 0.04,
    "Digital Wallet": 0.03,
    "Bank Transfer": 0.02
}).astype(float)

# High-value transactions have slightly higher failure risk
failure_probability += np.where(
    transactions["amount"] > 500,
    0.04,
    0
)

# Generate transaction status
random_values = np.random.random(NUM_TRANSACTIONS)

transactions["status"] = np.where(
    random_values < failure_probability,
    "Failed",
    "Successful"
)

transactions["status"].value_counts()
```

```python
failure_reasons = [
    "Insufficient Funds",
    "Authentication Failed",
    "Expired Card",
    "Technical Error",
    "Transaction Declined"
]

transactions["failure_reason"] = None

failed_mask = transactions["status"] == "Failed"

transactions.loc[failed_mask, "failure_reason"] = np.random.choice(
    failure_reasons,
    size=failed_mask.sum(),
    p=[0.30, 0.25, 0.15, 0.15, 0.15]
)

transactions[
    transactions["status"] == "Failed"
].head(10)
```

```python
total_transactions = len(transactions)

successful_transactions = (
    transactions["status"] == "Successful"
).sum()

failed_transactions = (
    transactions["status"] == "Failed"
).sum()

success_rate = (
    successful_transactions / total_transactions
) * 100

print(f"Total Transactions: {total_transactions:,}")
print(f"Successful Transactions: {successful_transactions:,}")
print(f"Failed Transactions: {failed_transactions:,}")
print(f"Success Rate: {success_rate:.2f}%")
```

```python
transactions = transactions.merge(
    customers,
    on="customer_id",
    how="left"
)

transactions.head()
```

```python
transactions = transactions.merge(
    merchants,
    on="merchant_id",
    how="left"
)

transactions.head()
```

```python
print("Dataset shape:", transactions.shape)

print("\nColumns:")
for column in transactions.columns:
    print("-", column)

print("\nMissing values:")
print(transactions.isnull().sum())
```

```python
print(
    "Duplicate transaction IDs:",
    transactions["transaction_id"].duplicated().sum()
)
```

```python
transactions["is_international"] = np.where(
    transactions["country"] != transactions["merchant_country"],
    True,
    False
)

transactions["is_international"].value_counts()
```

```python
transactions["is_international"].value_counts(normalize=True).mul(100).round(2)
```

```python
transactions["high_value_flag"] = (
    transactions["amount"] >= 500
)
```

```python
transactions = transactions.sort_values(
    ["customer_id", "transaction_timestamp"]
).reset_index(drop=True)
```

```python
customer_failures = (
    transactions
    .groupby("customer_id")["status"]
    .transform(lambda x: (x == "Failed").sum())
)

transactions["multiple_failures_flag"] = (
    customer_failures >= 3
)
```

```python
transactions["risk_flag"] = (
    transactions["high_value_flag"] |
    (
        transactions["multiple_failures_flag"] &
        transactions["is_international"]
    )
)
```

```python
print("High-value transactions:")
print(transactions["high_value_flag"].value_counts())

print("\nMultiple-failure transactions:")
print(transactions["multiple_failures_flag"].value_counts())

print("\nTransactions flagged for review:")
print(transactions["risk_flag"].value_counts())
```

```python
failed_without_reason = transactions[
    (transactions["status"] == "Failed") &
    (transactions["failure_reason"].isna())
]

successful_with_reason = transactions[
    (transactions["status"] == "Successful") &
    (transactions["failure_reason"].notna())
]

print("Failed transactions without reason:", len(failed_without_reason))
print("Successful transactions with failure reason:", len(successful_with_reason))
```
